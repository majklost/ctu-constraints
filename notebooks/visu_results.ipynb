```python
from matplotlib import pyplot as plt
from pathlib import Path
```

```python
results = Path("./outputs")
voxelmorph_results = results / "voxelmorph_test_generator"
```

```python
template = voxelmorph_results / "example_0_template_mask.png"
for i in range(10):
    deformed_mask = voxelmorph_results / f"example_{i}_deformed_mask.png"
    warped_template = voxelmorph_results / f"example_{i}_warped_mask.png"
    plt.figure(figsize=(15, 5))
    plt.subplot(1, 3, 1)
    plt.imshow(plt.imread(template), cmap="gray")
    plt.title("Template")
    plt.axis("off")
    plt.subplot(1, 3, 2)
    plt.imshow(plt.imread(deformed_mask), cmap="gray")
    plt.title(f"Deformed Mask {i}")
    plt.axis("off")
    plt.subplot(1, 3, 3)
    plt.imshow(plt.imread(warped_template), cmap="gray")
    plt.title(f"Warped Template {i}")
    plt.axis("off")
    if i==2:
        plt.tight_layout()
        # prepare for saving the figure and using in the report - tight layout and save
        # ensure the titles are not cut off and the figure is saved in the correct directory
    
        plt.savefig(voxelmorph_results / f"example_{i}_comparison.png", bbox_inches="tight", dpi=300)
        
    plt.show()
```

```python
twobranch_results = results / "affine_test_generator"
template = twobranch_results / "example_0_template.png"

for i in range(10):
    input_image = twobranch_results / f"example_{i}_input.png"
    
    mask = twobranch_results / f"example_{i}_mask.png"
    predicted_image = twobranch_results / f"example_{i}_pred.png"
    rotated_image = twobranch_results / f"example_{i}_rotated.png"
    plt.figure(figsize=(15, 10))
    plt.subplot(2, 3, 1)
    plt.imshow(plt.imread(template), cmap="gray")
    plt.title("Template")
    plt.axis("off")
    plt.subplot(2, 3, 2)
    print(plt.imread(input_image).shape)
    plt.imshow(plt.imread(input_image), cmap="gray")
    plt.title(f"Input Image {i}")
    plt.axis("off")
    plt.subplot(2, 3, 3)
    plt.imshow(plt.imread(mask), cmap="gray")
    plt.title(f"Mask {i}")
    plt.axis("off")
    plt.subplot(2, 3, 4)
    plt.imshow(plt.imread(predicted_image), cmap="gray")
    plt.title(f"Predicted Image {i}")
    plt.axis("off")
    plt.subplot(2, 3, 5)
    plt.imshow(plt.imread(rotated_image), cmap="gray")
    plt.title(f"Rotated Image {i}")
    plt.axis("off")
    plt.show()
```
