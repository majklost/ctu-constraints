```python
from constraints.models.affine import TwoBranch

torch.manual_seed(22)
model = TwoBranch()
model.train()

# Materialize LazyConv2d parameters before creating the optimizer.
with torch.no_grad():
    model(img.unsqueeze(0).float())

optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
loss1 = torch.nn.CrossEntropyLoss()
loss2 = CentroidLoss()

print("TASK:")
# show_torch_image(mask)
# show_torch_image(template)
print("RESULT:")
H,W = mask.shape[1], mask.shape[2]
for epoch in range(100):
    optimizer.zero_grad()
    decoded, angle, translation = model(img.unsqueeze(0).float())
    rotated_template = differentiable_rigid(template.float().unsqueeze(0), angle, dx=translation[:, 0], dy=translation[:, 1])
    print("angle:", rad2deg(angle).item())
    loss1_val = loss1(decoded, mask_target.unsqueeze(0))
    loss2_val =  100 * loss2(mask.float().unsqueeze(0), rotated_template.unsqueeze(0))
    loss = loss1_val + loss2_val
    loss.backward()
    
    optimizer.step()
    if epoch % 10 == 0:
        show_torch_image(torch.argmax(decoded[0], dim=0))
        show_torch_image(rotated_template)
        
    print(f"Epoch {epoch}, Losses: {loss1_val.item()} {loss2_val.item()}")
```

## Deform template

```python
from constraints.voxelmorph.utils import random_disp,spatial_transform
from constraints.visu.helpers import show_torch_image
```

```python
img_b = img.unsqueeze(0).float()
mask_b = mask.unsqueeze(0).float()
torch.manual_seed(23)
test_field = random_disp(img_b.shape,magnitude=2.0,integrations=5,fractal_mode="upsample")
print(test_field.shape)
test_magnitude = torch.sqrt(torch.sum(test_field**2, dim=1))
show_torch_image(test_magnitude[0], title="Random Displacement Magnitude", cmap="hot")
warped_img = spatial_transform(img_b, test_field)
warped_mask = spatial_transform(mask_b, test_field, method="nearest")
shapes = (img_b.shape, warped_img.shape, mask_b.shape, warped_mask.shape)
print("Shapes:", shapes)
show_torch_image(img_b[0], title="Original Image", cmap="gray")
show_torch_image(warped_img[0], title="Warped Image", cmap="gray")
show_torch_image(mask_b[0], title="Original Mask", cmap="gray")
show_torch_image(warped_mask[0], title="Warped Mask", cmap="gray")

# create 2x2 grid random displacement, warped image, mask (named template), warped mask
def torch_to_numpy(tensor):
    return tensor.permute(1, 2, 0).detach().cpu().numpy() if tensor.ndim == 3 else tensor.detach().cpu().numpy()
fig, axs = plt.subplots(2, 2, figsize=(10, 10))
axs[0, 0].imshow(torch_to_numpy(test_magnitude[0]), cmap='hot')
axs[0, 0].set_title("Random Displacement Magnitude")
axs[0, 1].imshow(torch_to_numpy(warped_img[0]), cmap='gray')
axs[0, 1].set_title("Warped Image")
axs[1, 0].imshow(torch_to_numpy(mask_b[0]), cmap='gray')
axs[1, 0].set_title("Template")
axs[1, 1].imshow(torch_to_numpy(warped_mask[0]), cmap='gray')
axs[1, 1].set_title("Mask")
for ax in axs.flat:
    ax.axis('off')
plt.tight_layout()
plt.savefig("warped_results.png")
```
